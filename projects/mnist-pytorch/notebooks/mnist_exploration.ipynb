# MNIST data exploration

A small reproducible walkthrough of the dataset used by this project: sample images, class balance, and the normalized input shape.

```python
from collections import Counter
from pathlib import Path
import sys

import matplotlib.pyplot as plt

candidates = [Path.cwd(), Path.cwd().parent, Path.cwd() / "projects/mnist-pytorch"]
PROJECT_DIR = next(path for path in candidates if (path / "src").is_dir())
sys.path.insert(0, str(PROJECT_DIR))
from src.data import load_dataset

train_dataset = load_dataset(train=True, data_dir=PROJECT_DIR / "data", download=True)
print(f"Training examples: {len(train_dataset):,}")
```

```python
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for index, axis in enumerate(axes.flat):
    image, label = train_dataset[index]
    axis.imshow(image.squeeze(), cmap="gray")
    axis.set_title(f"Label: {label}")
    axis.axis("off")
plt.tight_layout()
```

```python
class_counts = Counter(train_dataset.targets.tolist())
plt.bar(class_counts.keys(), class_counts.values())
plt.xticks(range(10))
plt.xlabel("Digit")
plt.ylabel("Number of training samples")
plt.title("MNIST training class distribution")
```

```python
sample, label = train_dataset[0]
print(f"Tensor shape: {tuple(sample.shape)}")
print(f"Normalized range: [{sample.min():.2f}, {sample.max():.2f}]")
print(f"Label: {label}")
```
